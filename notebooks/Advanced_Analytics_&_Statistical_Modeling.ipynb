## **Confidence Interval**

```python
import scipy.stats as stats
import numpy as np
```

```python
confidence = 0.95

monthly_charge = df['Monthly Charge']

mean = np.mean(monthly_charge)
std_error = stats.sem(monthly_charge)

confidence_interval = stats.t.interval(
    confidence=confidence,
    df=len(monthly_charge)-1,
    loc=mean,
    scale=std_error
)

print("Mean Monthly Charge:", round(mean,2))
print("95% Confidence Interval:", confidence_interval)
```

**Business Interpretation**

The average monthly charge paid by customers is $64.76.

# **Hypothesis Testing**

Independent Sample t-**Test**

```python
from scipy.stats import ttest_ind
```

```python
churn_yes = df[df['Churn Label'] == 'Yes']['Monthly Charge']

churn_no = df[df['Churn Label'] == 'No']['Monthly Charge']
```

```python
t_stat, p_value = ttest_ind(churn_yes, churn_no)

print("T-Statistic :", t_stat)
print("P-Value :", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a significant difference in Monthly Charges.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no significant difference.")
```

## **Business Insight&Recommendation**
Independent Sample t-Test

An independent sample t-test was conducted to determine whether there was a significant difference in monthly charges between churned and retained customers.

The test produced a t-statistic of 16.54 and a p-value of 2.71 × 10⁻⁶⁰, which is substantially below the significance level of 0.05. Therefore, the null hypothesis was rejected.

The findings indicate that customers who churn have significantly different monthly charges than those who remain with the company. This suggests that pricing is an important factor influencing customer retention and should be considered when developing retention strategies.

The company should identify customers with relatively high monthly charges and consider:

* Personalized discounts
* Loyalty rewards
* Flexible pricing plans
* Bundled service offers
* Proactive customer engagement

These strategies may help reduce churn among high-paying customers.

**One-Way ANOVA (Analysis of Variance)**

```python
from scipy.stats import f_oneway
```

```python
month_to_month = df[df['Contract'] == 'Month-to-Month']['Monthly Charge']

one_year = df[df['Contract'] == 'One Year']['Monthly Charge']

two_year = df[df['Contract'] == 'Two Year']['Monthly Charge']
```

## **Customer Segmentation (Clustering)**

## Objective

Customer Segmentation is an unsupervised machine learning technique used to group customers with similar characteristics. This enables businesses to understand customer behavior, improve marketing strategies, personalize services, and optimize customer retention efforts.

In this section, we will:

- Select important customer features
- Standardize the data
- Determine the optimal number of clusters using the Elbow Method and Silhouette Score
- Build a K-Means clustering model
- Visualize customer segments
- Analyze each cluster
- Generate business insights and recommendations

```python
# Select features for clustering

cluster_features = [
    'Age',
    'Tenure in Months',
    'Monthly Charge',
    'Total Revenue',
    'Satisfaction Score',
    'Number of Referrals'
]

X = df[cluster_features]

X.head()
```

```python
X.isnull().sum()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Shape:", X_scaled.shape)
```

```python
scaled_df = pd.DataFrame(X_scaled, columns=cluster_features)

scaled_df.head()
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Calculate WCSS for different values of K
wcss = []

for k in range(1, 11):
    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plot the Elbow Curve
plt.figure(figsize=(8,5))
plt.plot(range(1, 11), wcss, marker='o', linewidth=2)
plt.title('Elbow Method for Optimal Number of Clusters', fontsize=14)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Inertia)')
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

# Display scores
for k, score in zip(range(2, 11), silhouette_scores):
    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
plt.figure(figsize=(8,5))
plt.plot(range(2, 11), silhouette_scores, marker='o', color='green')
plt.title('Silhouette Score for Different K Values', fontsize=14)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()
```

```python
# Final KMeans Model

kmeans = KMeans(
    n_clusters=4,
    init='k-means++',
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

# Add cluster labels to original dataframe
df["Cluster"] = clusters

df.head()
```

```python
cluster_percent = (
    df["Cluster"]
      .value_counts(normalize=True)
      .sort_index() * 100
)

cluster_percent
```

```python
from sklearn.decomposition import PCA

# Reduce dimensions
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

# Create dataframe
pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

pca_df["Cluster"] = clusters
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,7))

for cluster in sorted(pca_df["Cluster"].unique()):
    subset = pca_df[pca_df["Cluster"] == cluster]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=f"Cluster {cluster}",
        alpha=0.7
    )

plt.title("Customer Segmentation using PCA + KMeans", fontsize=16)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.grid(True)

plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

model_df["Churn Label"] = LabelEncoder().fit_transform(model_df["Churn Label"])

model_df["Churn Label"].value_counts()
```

```python
categorical_cols = model_df.select_dtypes(include="object").columns

print(categorical_cols)
```

# **🤖 Model 1: Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression

# Create the model
log_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
log_model.fit(X_train_scaled, y_train)
```

```python
# Predict class labels
y_pred_log = log_model.predict(X_test_scaled)

# Predict probabilities
y_prob_log = log_model.predict_proba(X_test_scaled)[:, 1]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("Logistic Regression Performance")
print("-"*45)

print(f"Accuracy : {accuracy_score(y_test, y_pred_log):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_log):.4f}")
print(f"Recall   : {recall_score(y_test, y_pred_log):.4f}")
print(f"F1 Score : {f1_score(y_test, y_pred_log):.4f}")
print(f"ROC AUC  : {roc_auc_score(y_test, y_prob_log):.4f}")
```

**Business Insight:**  
Organizations can proactively identify customers at risk and implement targeted retention strategies before churn occurs.
